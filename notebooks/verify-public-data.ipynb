# Verify the public OSM-only dataset
Python 3.11+, standard library only for these checks. Full construction history is in docs/. No historical database is required. Cells were executed sequentially in Python with saved outputs, not via a Jupyter kernel.

```python
from pathlib import Path
import json, subprocess, sys
root = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'scripts/verify_dataset.py').exists()), None)
if root is None:
    raise RuntimeError('Open this notebook inside the repository checkout')
release = json.loads((root / 'data/release.json').read_text())
manifest = json.loads((root / 'data/manifest.json').read_text())
print('Dataset:', release['dataset_id'])
print('OSM snapshot:', manifest['data_timestamp_utc'])
print('Public layers:', manifest['layers'])
```

```python
completed = subprocess.run([sys.executable, str(root / 'scripts/verify_dataset.py')], cwd=root, capture_output=True, text=True, check=True)
print(completed.stdout)
```

```python
audit = json.loads((root / 'docs/audits/public-distribution-verification.json').read_text())
assert audit['status'] == 'pass'
assert audit['pbf_unchanged'] and audit['freelist_pages'] == 0
assert len(audit['layers']) == 8
print('Recorded read-back audit:', audit['audit_utc'])
print('Every retained layer had unchanged, valid, nonempty geometry.')
print('This cell reads the recorded geometry audit; the previous cell reruns archive verification.')
```
